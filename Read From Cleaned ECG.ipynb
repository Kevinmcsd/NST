```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import os
import random
```

```python
wd='Cleaned ECG'
File_Names=os.listdir(wd)
Beats=[]
Names=[]
for File_Name in File_Names:
    img = cv2.imread('Cleaned ECG/'+File_Name,0)
    pix_found=[]
    max_width=img.shape[0]
    height=img.shape[1]
    for col in range(height):
        unique, counts = np.unique(img[:,col], return_counts=True)
        T=np.where(img[:,col]==unique[0])
        if unique[0]<=150:
            Beat=np.floor(160*(T[0][0])/max_width)
            pix_found.append(210-Beat)
        else:
            pix_found.append(np.nan)
    Temp_beat=pd.Series(pix_found)
    Beat=Temp_beat.interpolate('linear')
    Beats.append(dict(zip(range(height),Beat)))
    Names.append(File_Name.split('.')[0])
    df=pd.DataFrame({"File_Names":Names,"Beats_Dict":Beats})
```

```python
df.to_csv("Test_ECG.csv")
```
