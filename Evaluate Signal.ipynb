```python
import pandas as pd
import numpy as np
import json
import ast
```

```python
df=pd.read_csv('Test_ECG.csv')
```

```python
index=3
print(df.iloc[index,1])
Dict = ast.literal_eval(df.iloc[index,2])
ECG_net=list(Dict.values())
Base_line=np.median(ECG_net)
Base_line
```

```python
count=0
occurance=0
for beat in ECG_net:
    if beat>Base_line+5:
         count=count+1
    else:
        if count>=20:
            occurance=occurance+1
        count=0
print(occurance)
```
