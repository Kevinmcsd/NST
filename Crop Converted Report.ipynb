```python
import matplotlib.pyplot as plt
import numpy as np
import os
```

```python
wd='Data\\Converted Reports'
File_Names=os.listdir(wd)
for File_Name in File_Names:
    Report=plt.imread('Data/Converted Reports/'+File_Name)
    Cropped_Report=Report[
                   np.where(Report[:,0,0] < 250)[0][0]:
                   np.where(Report[:,0,0] < 250)[0][-1],
                   0:Report.shape[1],:]
    plt.imsave('Data/Cropped Reports/'+File_Name,Cropped_Report)
```
