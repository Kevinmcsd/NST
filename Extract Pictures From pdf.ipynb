```python
from pdf2image import convert_from_path
import os
```

```python
wd='Data\\nst1-75'
File_Names=os.listdir(wd)

for File_Name in File_Names:
    images = convert_from_path('Data\\nst1-75\\'+File_Name)

    images[0].save('Data\\Converted Reports\\'+File_Name.split('.')[0]+'.jpg', 'JPEG')
```
